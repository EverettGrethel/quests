```python
import numpy as np
import zipfile
from pathlib import Path

def find_embeddings_file(directory, model, dataset):
    directory = Path(directory)

    npy_dir = directory / "npy"
    if npy_dir.exists():
        for file in npy_dir.iterdir():
            if file.is_file():
                name = file.name
                if name.startswith(model) and name.endswith(f"_{dataset}.npy"):
                    return file.resolve()
                
    npz_dir = directory / "npz"
    if npz_dir.exists():
        for file in npz_dir.iterdir():
            if file.is_file():
                name = file.name
                if name.startswith(model) and name.endswith(f"_{dataset}.npz"):
                    return file.resolve()

    raise FileNotFoundError(
        f"Could not find embeddings (.npy or .npz) for model {model} "
        f"and dataset {dataset} in directories {directory}/npy or {directory}/npz"
    )

def is_valid_npz(path: Path) -> bool:
    try:
        with zipfile.ZipFile(path) as z:
            return z.testzip() is None
    except Exception:
        return False

def validate_object_embeddings(obj) -> bool:
    if not isinstance(obj, np.ndarray):
        return False
    if obj.dtype != object:
        return False
    if obj.size == 0:
        return False
    if not isinstance(obj[0], np.ndarray):
        return False
    return True

def validate_numeric_embeddings(X) -> bool:
    if not isinstance(X, np.ndarray):
        return False
    if X.dtype != np.float32:
        return False
    if X.ndim < 3:
        return False
    if not np.isfinite(X).all():
        return False
    return True

models = [
    # "mace_mp_small",
    # "mace_mp_medium",
    # "mace_mp_large",
    # "mace_off_small",
    # "mace_off_medium",
    # "mace_off_large",
    # "uma-s-1p1",
    # "uma-m-1p1",
    "eqV2_31M_omat_mp_salex",
    "eqV2_86M_omat_mp_salex",
    "eqV2_153M_omat_mp_salex",
    "eqV2_dens_31M_mp",
    "eqV2_dens_86M_mp",
    "eqV2_dens_153M_mp",
    "orb-v3-conservative-inf-omat",
    "orb-v3-conservative-20-omat",
    "orb-v3-conservative-inf-mpa",
    "orb-v3-conservative-20-mpa",
]

datasets = [
    "Graphene",
    "Diamond",
    "Graphite",
    "Nanotubes",
    "Fullerenes",
    "Liquid",
]

base_dir = Path("/data/grethel/embeddings/embeddings_raw/npz")
out_dir = base_dir / "converted_npy"
out_dir.mkdir(exist_ok=True)

failed = []

for model in models:
    for dataset in datasets:
        try:
            npz_path = find_embeddings_file(base_dir, model, dataset)
            print(f"Processing {npz_path.name}")

            # Quick ZIP integrity check
            if not is_valid_npz(npz_path):
                raise RuntimeError("Invalid or corrupted NPZ archive")

            with np.load(npz_path, allow_pickle=True) as f:
                if "embeddings" not in f:
                    raise KeyError("Missing 'embeddings' key")
                obj = f["embeddings"]

            if not validate_object_embeddings(obj):
                raise RuntimeError("Invalid object embeddings structure")

            # Collapse object nesting
            X = np.stack(obj).astype(np.float32)

            if not validate_numeric_embeddings(X):
                raise RuntimeError("Invalid numeric embeddings after stacking")

            out_path = out_dir / (npz_path.stem + ".npy")
            np.save(out_path, X)

            print(f"  → saved {out_path} | shape={X.shape}")

        except Exception as e:
            print(f"  ✗ FAILED: {model} / {dataset} → {e}")
            failed.append(
                {
                    "model": model,
                    "dataset": dataset,
                    "path": str(npz_path) if "npz_path" in locals() else None,
                    "error": str(e),
                }
            )

print("\n========== SUMMARY ==========")
print(f"Total failed files: {len(failed)}")

for f in failed:
    print(f"- {f['model']} / {f['dataset']} : {f['error']}")
```
